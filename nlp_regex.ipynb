```python
import re
```

```python
text="""Born	Jeffrey Preston Jorgensen
January 12, 1964 (age 62)
Albuquerque, New Mexico, U.S.
Education	Princeton University (BSE)
Occupations	
Businessmanmedia proprietorinvestor
Known for	Founding Amazon
Title	
Founder and executive chairman of Amazon
Founder of Blue Origin
Owner of The Washington Post
Founder of Bezos Expeditions
Executive chair of Bezos Earth Fund
Founder of Bezos Academy
"""
```

```python
def pattern_match(pattern,text):
    match=re.findall(pattern,text)
    return match[0]
```

```python
def personal_info(text):
    age=pattern_match('age (\d+)',text)
    name=pattern_match('Born(.*)\n',text)
    birth_place=pattern_match('\(age.*\n(.*)',text)
    birth_date=pattern_match('\Born.*\n(.*)',text)

    return{
        'age':int(age),
        'name':name.strip(),
        'birthdate':birth_date.strip(),
        'birthplace':birth_place.strip()
    }
```

```python
personal_info(text)
```
